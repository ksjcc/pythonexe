任务1：score_level(score)

```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError(f"分数必须在 0~100 之间，当前为 {score}")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for x in [100, 95, 90, 85, 80, 70, 60, 50, 0]:
    print(x, "->", score_level(x))

try:
    score_level(120)
except ValueError as e:
    print("异常：", e)

try:
    score_level(-5)
except ValueError as e:
    print("异常：", e)
```

任务3：递归版与迭代版阶乘

```python
def fact_rec(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("输入必须是非负整数")
    if n == 0:
        return 1
    return n * fact_rec(n - 1)

def fact_iter(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("输入必须是非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result

for n in [0, 1, 5, 10]:
    print(f"n={n:2d}  递归={fact_rec(n):<10} 迭代={fact_iter(n):<10} 相等={fact_rec(n) == fact_iter(n)}")
```

任务4：summary(values) 返回元组并解包

```python
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    return min(values), max(values), sum(values) / len(values)

result = summary([85, 58, 92, 76, 60, 89])
print("返回元组：", result)

mn, mx, avg = summary([85, 58, 92, 76, 60, 89])   # 元组解包
print(f"最小值={mn}, 最大值={mx}, 平均值={avg:.2f}")
```

任务3：parse_student() 校验单行学生记录

```python
def parse_student(line):
    '''将一行学生记录解析为 (name, score)，无效记录抛出 ValueError。'''
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量不为 2（当前 {len(parts)} 个字段）")
    name, score_str = parts
    if name == "":
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩 '{score_str}' 不是整数") from None
    if not (0 <= score <= 100):
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    return name, score

# 测试典型输入：有效 / 成绩非数字 / 成绩越界 / 字段不足 / 姓名为空
for line in ["Alice,85", "Grace,abc", "Henry,105", "Incomplete", ",90"]:
    try:
        print(f"{line!r:15s} ->", parse_student(line))
    except ValueError as e:
        print(f"{line!r:15s} -> ValueError: {e}")
```

任务4：按行处理文件，捕获无效记录异常并继续

```python
valid_records = []
invalid_log = []

with open(students_path, "r", encoding="utf-8") as f:
    lines = f.read().splitlines()

# 跳过表头，行号从 2 开始计
for lineno, line in enumerate(lines[1:], start=2):
    try:
        name, score = parse_student(line)
        valid_records.append({"name": name, "score": score})
    except ValueError as e:
        invalid_log.append((lineno, line, str(e)))   # 记录行号与原因，继续处理

print("有效记录：")
for r in valid_records:
    print("  ", r)

print("无效记录（行号 / 原文 / 原因）：")
for lineno, line, reason in invalid_log:
    print(f"   第{lineno}行 {line!r}: {reason}")
```

任务5：统计结果写入 report.txt（演示 w 与 a 模式）

```python
count_valid = len(valid_records)
count_invalid = len(invalid_log)
avg_score = sum(r["score"] for r in valid_records) / count_valid
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / count_valid
print(f"有效={count_valid} 无效={count_invalid} 平均分={avg_score:.2f} 及格率={pass_rate:.1%}")

report_path = os.path.join(data_dir, "report.txt")

# w 模式：覆盖写入（原内容被清空）
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数：{count_valid}\n")
    f.write(f"无效记录数：{count_invalid}\n")
    f.write(f"平均分：{avg_score:.2f}\n")
    f.write(f"及格率：{pass_rate:.1%}\n")
print("w 模式写入后 report.txt 内容：")
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())

# a 模式：追加写入（已有内容保留）
with open(report_path, "a", encoding="utf-8") as f:
    f.write("(追加示例：a 模式不会覆盖已有内容)\n")
print("a 模式追加后 report.txt 内容：")
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())
```

实验5：类、继承与对象复制

任务1：定义 Student 类（is_passed / __str__）

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

# 简单测试
s = Student("Alice", 85)
print(s, "| 及格：", s.is_passed())
print("Student('Bob', 58) 及格：", Student("Bob", 58).is_passed())
```

任务2：用实验4的有效记录创建 Student 对象

```python
# 使用实验4解析出的每条有效记录创建对象
students = [Student(r["name"], r["score"]) for r in valid_records]

for stu in students:
    print(stu, "| 及格：", stu.is_passed())
```

任务3：GraduateStudent 继承 Student 并重写 __str__

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)          # 初始化父类属性
        self.research_topic = research_topic   # 新增属性

    def __str__(self):                          # 重写 __str__
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic})")

gs = GraduateStudent("Eva", 60, "自然语言处理")
print(gs)
print("继承检查 isinstance(gs, Student)：", isinstance(gs, Student))
print("调用父类方法 is_passed()：", gs.is_passed())
```

任务4：直接赋值 / copy.copy() / copy.deepcopy() 对比

```python
import copy

def make_original():
    '''每次比较都从新建的原始对象开始。'''
    return {"name": "Alice", "scores": [85, 90]}

# 1) 直接赋值：两个名字指向同一个对象
orig = make_original()
alias = orig
alias["scores"][0] = 0
print("直接赋值：alias 是 orig 的同一对象 ->", orig is alias, "| orig:", orig)

# 2) copy.copy()：浅拷贝，内部列表仍然共享
orig = make_original()
shallow = copy.copy(orig)
shallow["scores"][0] = 0
print("浅拷贝：修改 shallow 的内部列表后 orig:", orig)
print("        内部列表是否共享：", orig["scores"] is shallow["scores"])

# 3) copy.deepcopy()：深拷贝，内部列表完全独立
orig = make_original()
deep = copy.deepcopy(orig)
deep["scores"][0] = 0
print("深拷贝：修改 deep 的内部列表后 orig:", orig)
print("        内部列表是否独立：", orig["scores"] is deep["scores"])
```
