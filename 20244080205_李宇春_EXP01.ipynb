实验2：函数、条件、循环与递归

```python
任务1：score_level(score)
```

```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError(f"分数必须在 0~100 之间，当前为 {score}")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for x in [100, 95, 90, 85, 80, 70, 60, 50, 0]:
    print(x, "->", score_level(x))

try:
    score_level(120)
except ValueError as e:
    print("异常：", e)

try:
    score_level(-5)
except ValueError as e:
    print("异常：", e)
```

```python
任务2：for 与 while 计算 1~100 整数和
```

```python
# for 实现
total_for = 0
for i in range(1, 101):
    total_for += i

# while 实现
total_while = 0
j = 1
while j <= 100:
    total_while += j
    j += 1

print("for   结果：", total_for)
print("while 结果：", total_while)
print("两者相等：", total_for == total_while)
```

```python
任务3：递归版与迭代版阶乘
```

```python
def fact_rec(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("输入必须是非负整数")
    if n == 0:
        return 1
    return n * fact_rec(n - 1)

def fact_iter(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("输入必须是非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result

for n in [0, 1, 5, 10]:
    print(f"n={n:2d}  递归={fact_rec(n):<10} 迭代={fact_iter(n):<10} 相等={fact_rec(n) == fact_iter(n)}")
```

```python
任务4：summary(values) 返回元组并解包
```

```python
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    return min(values), max(values), sum(values) / len(values)

result = summary([85, 58, 92, 76, 60, 89])
print("返回元组：", result)

mn, mx, avg = summary([85, 58, 92, 76, 60, 89])   # 元组解包
print(f"最小值={mn}, 最大值={mx}, 平均值={avg:.2f}")
```

```python
实验3：列表、字典、元组与基础统计
```

```python
names  = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]
print(list(zip(names, scores)))
```

```python
任务1：zip() 构建字典，items() 逐项输出
```

```python
score_dict = dict(zip(names, scores))
print(score_dict)

for name, score in score_dict.items():
    print(f"{name}: {score}")
```

```python
任务2：enumerate() 输出序号、姓名、成绩，序号从 1 开始
```

```python
for idx, (name, score) in enumerate(zip(names, scores), start=1):
    print(idx, name, score)
```

```python
任务3：列表推导式提取及格姓名、生成成绩平方列表
```

```python
passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]

print("及格学生：", passed_names)
print("成绩平方：", squared_scores)
```

```python
任务4：统计人数、平均分、最高分、最低分、及格率
```

```python
count = len(scores)
avg = sum(scores) / count
max_score = max(scores)
min_score = min(scores)
pass_count = sum(1 for s in scores if s >= 60)
pass_rate = pass_count / count

print(f"人数：{count}")
print(f"平均分：{avg:.2f}")
print(f"最高分：{max_score}")
print(f"最低分：{min_score}")
print(f"及格人数：{pass_count}")
print(f"及格率：{pass_rate:.1%}")
```

```python
任务5：alias 与 copy 对比（用单独演示列表，不改原始 scores）
```

```python
demo = [10, 20, 30]
alias = demo
copied = demo.copy()

print("修改前：demo =", demo, " alias =", alias, " copied =", copied)

alias[0] = 999

print("修改 alias[0] 后：")
print("demo   =", demo)      # [999, 20, 30]，被 alias 影响
print("alias  =", alias)     # [999, 20, 30]
print("copied =", copied)    # [10, 20, 30]，不受影响
```
